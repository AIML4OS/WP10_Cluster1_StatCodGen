```python
word_lengths_synth = [len(word_synth) for word_synth in words_synth]
length_counts_synth = pd.Series(word_lengths_synth).value_counts().sort_index()

word_lengths_synth_wz08 = [len(word_synth_wz08) for word_synth_wz08 in words_synth_wz08]
length_counts_synth_wz08 = pd.Series(word_lengths_synth_wz08).value_counts().sort_index()

word_lengths_train = [len(word_train) for word_train in words_train]
length_counts_train = pd.Series(word_lengths_train).value_counts().sort_index()

# Filter für train, um nur Wortlängen bis 50 zu behalten
length_counts_train = length_counts_train[length_counts_train.index <= 50]

# Beide Häufigkeiten nach Länge sortieren
length_counts_sorted_synth = length_counts_synth.sort_values(ascending=False)
length_counts_sorted_synth_wz08 = length_counts_synth_wz08.sort_values(ascending=False)
length_counts_sorted_train = length_counts_train.sort_values(ascending=False)

# Plot
plt.figure(figsize=(12, 6))

# Säulen für die ersten Wortlängen
bar_width = 0.35  # Breite der Säulen
index = length_counts_sorted_synth.index
plt.bar(index - bar_width/2, length_counts_sorted_synth.values, bar_width, label='words_synth', color='orange', alpha=0.7)

# Säulen für die zweiten Wortlängen
index_wz08 = length_counts_sorted_synth_wz08.index
plt.bar(index_wz08 + bar_width/2, length_counts_sorted_synth_wz08.values, bar_width, label='words_synth_wz08', color='blue', alpha=0.7)

# Säulen für die dritten Wortlängen (nur bis 50)
index_train = length_counts_sorted_train.index
plt.bar(index_train + bar_width/2, length_counts_sorted_train.values, bar_width, label='words_train (max 50)', color='green', alpha=0.7)

# Diagramm anpassen
plt.title('Vergleich der Häufigkeit der Wortlängen (bis Länge 50)')
plt.xlabel('Wortlänge (Anzahl der Zeichen)')
plt.ylabel('Häufigkeit')
plt.xticks(rotation=45)
plt.tight_layout()
plt.legend()
plt.show()
```

```python
plt.figure(figsize=(12, 6))
bar_width = 0.25  # Breite der Balken

# Alle Wortlängen bis max Länge
all_lengths = sorted(set(length_counts_synth.index)
                     | set(length_counts_synth_wz08.index)
                     | set(length_counts_train.index))

# Relative Häufigkeiten berechnen
rel_synth = [length_counts_synth.get(x, 0) / length_counts_synth.sum() for x in all_lengths]
rel_synth_wz08 = [length_counts_synth_wz08.get(x, 0) / length_counts_synth_wz08.sum() for x in all_lengths]
rel_train = [length_counts_train.get(x, 0) / length_counts_train.sum() for x in all_lengths]

# Balken für words_synth
plt.bar([x - bar_width for x in all_lengths], rel_synth, 
        width=bar_width, label='words_synth', color='orange', alpha=0.7)

# Balken für words_synth_wz08
plt.bar(all_lengths, rel_synth_wz08, 
        width=bar_width, label='words_synth_wz08', color='blue', alpha=0.7)

# Balken für words_train
plt.bar([x + bar_width for x in all_lengths], rel_train, 
        width=bar_width, label='words_train (max 50)', color='green', alpha=0.7)

# Diagramm anpassen
plt.title('Vergleich der relativen Häufigkeit der Wortlängen (bis Länge 50)')
plt.xlabel('Wortlänge (Anzahl der Zeichen)')
plt.ylabel('Relative Häufigkeit')
plt.xticks(all_lengths, rotation=45)
plt.tight_layout()
plt.legend()
plt.show()
```

## Anteil der häufigsten 10 Wörter

#### Synthetische Daten WZ25
#### Synthetic Data WZ25

```python
total_words_synth = sum(count_synth for word_synth, count_synth in word_counts_synth.items())
word_perc_synth = {word_synth: count_synth / total_words_synth * 100 for word_synth, count_synth in word_counts_synth.most_common(10)}

other_count = sum(count_synth for word_synth, count_synth in word_counts_synth.items() if word_synth not in word_perc_synth)
word_perc_synth["Sonstiges"] = other_count / total_words_synth * 100

labels_synth = list(word_perc_synth.keys())
values_synth = list(word_perc_synth.values())

colors_synth = plt.cm.Paired.colors[:len(labels_synth)]
fig, ax = plt.subplots(figsize=(12, 8))

# Kreisdiagramm links ohne Labels
ax_pie = fig.add_axes([0.1, 0.1, 0.35, 0.8])
ax_pie.pie(values_synth, startangle=90, colors=colors_synth, wedgeprops={'edgecolor': 'black'})
ax_pie.set_title('Anteil der häufigsten Wörter', fontsize=16)

# Tabelle rechts mit Prozentwerten und Farben
ax_table = fig.add_axes([0.6, 0.1, 0.35, 0.8])

# Erstellen der Tabellenzeilen
table_data = [
    (label, f'{value:.1f}%')
    for label, value in zip(labels_synth, values_synth)
]
table = ax_table.table(cellText=table_data, colLabels=['Wort', 'Prozent'], cellLoc='center', loc='center')
table.auto_set_font_size(False)
table.set_fontsize(12)
table.scale(1.2, 1.2)

for i, (label, value) in enumerate(table_data):
    table[(i+1, 0)].set_facecolor(colors_synth[i])
    table[(i+1, 1)].set_facecolor(colors_synth[i])

ax_table.axis('off')
plt.tight_layout()
plt.show()
```

```python
total_words_synth = sum(count_synth for word_synth, count_synth in word_counts_synth.items())
word_perc_synth = {word_synth: count_synth / total_words_synth * 100 for word_synth, count_synth in word_counts_synth.most_common(10)}

labels_synth = list(word_perc_synth.keys())
values_synth = list(word_perc_synth.values())

plt.figure(figsize=(6, 6))
plt.pie(values_synth, labels=labels_synth, autopct='%1.1f%%', startangle=90, colors=plt.cm.Paired.colors)

plt.title('Anteil der häufigsten Wörter')
plt.show()
```

# Tokenanzahl

#### Mann Whithney U-Test
Ziel: Vergleich von zwei unabhängigen Stichproben, ob sie aus der gleichen Verteilung stammen (oft: Unterschiede in Mediane oder Rangverteilungen).
##### Interpretation
Nullhypothese H0: Die beiden Gruppen stammen aus der gleichen Verteilung.
Alternativhypothese H1: Die Gruppen unterscheiden sich in der Verteilung (z. B. Median).
Signifikanz: Wird der kritische Wert (oder p-Wert) überschritten, wird H0 verworfen.

```python
def token_lengths(text_data):
    return [len(word) for text in text_data for word in re.findall(r'\b\w+\b', text.lower())]
```

```python
train_token_lengths = token_lengths(train_data_filtered['Taetigkeit'])
synthetic_token_lengths = token_lengths(synth_data['Generierte Daten'])
stat, p_value = stats.mannwhitneyu(train_token_lengths, synthetic_token_lengths, alternative='two-sided')

print(f"U-Wert: {stat}")
print(f"p-Wert: {p_value}")

if p_value < 0.05:
    print("Es gibt einen signifikanten Unterschied in der Tokenlänge zwischen Trainingsdaten und synthetischen Daten.")
else:
    print("Es gibt keinen signifikanten Unterschied in der Tokenlänge zwischen Trainingsdaten und synthetischen Daten.")

plt.boxplot([train_token_lengths, synthetic_token_lengths], labels=['Trainingsdaten', 'Synthetische Daten'])
plt.title('Vergleich der Tokenlängen')
plt.ylabel('Tokenlängen')
plt.show()
```

```python
def get_top_n_token_stats(text_data, n=15):
    """
    Gibt die n längsten Texte nach Tokenanzahl zurück,
    inklusive Tokenliste und Zeichenanzahl pro Token.
    
    Parameter:
        text_data (Liste oder Series von Strings): Die Texte.
        n (int): Anzahl der längsten Texte.
    
    Rückgabe:
        Liste von Dictionaries:
            {
                'tokens': Liste von Tokens,
                'num_tokens': Anzahl Tokens,
                'token_lengths': Liste der Zeichenlängen pro Token
            }
    """
    result = []
    
    for text in text_data:
        tokens = text.split()
        token_lengths = [len(token) for token in tokens]
        result.append({
            'tokens': tokens,
            'num_tokens': len(tokens),
            'token_lengths': token_lengths
        })
    result.sort(key=lambda x: x['num_tokens'], reverse=True)
    return result[:n]
```

# Cosinus similarity

Kosinus-Ähnlichkeit (englisch cosine similarity) ist eine Metrik in der Data Science und im maschinellen Lernen, um die Ähnlichkeit zwischen zwei Daten in einem mehrdimensionalen Raum zu messen.

Wertebereich: Die Kosinus-Ähnlichkeit liegt typischerweise zwischen -1 und 1.
1: Vektoren zeigen in exakt dieselbe Richtung (höchste Ähnlichkeit).
0: Vektoren sind orthogonal (im rechten Winkel), d.h. es besteht keine inhaltliche Ähnlichkeit.
-1: Vektoren zeigen in exakt entgegengesetzte Richtungen (maximale Unähnlichkeit).

```python
merged_data = pd.merge(train_data_filtered_grouped, synth_data_grouped_wz08, left_on='WZ', right_on='wz_code', how='inner')

# Berechnung der Kosinus-Ähnlichkeiten
cosine_similarities = []
vectorizer = TfidfVectorizer(stop_words=stop_words_de)

for index, row in merged_data.iterrows():
    taetigkeit_text = row['Taetigkeit']
    generierte_daten_text = row['Generierte Daten']
    tfidf_matrix = vectorizer.fit_transform([taetigkeit_text, generierte_daten_text])
    cosine_sim = cosine_similarity(tfidf_matrix[0:1], tfidf_matrix[1:2])
    cosine_similarities.append({
        'WZ': row['WZ'],
        'Titel': row['Titel'],
        'Kosinus-Ähnlichkeit': cosine_sim[0][0]
    })

cosine_sim_df = pd.DataFrame(cosine_similarities)
cosine_sim_df.head()
```

```python
data = cosine_sim_df['Kosinus-Ähnlichkeit']

# Berechnung der Quartile und IQR
q1 = np.percentile(data, 25)
median = np.percentile(data, 50)
q3 = np.percentile(data, 75)
iqr = q3 - q1

# Bestimmung von Whisker (ohne Ausreißer)
lower_whisker = q1 - 1.5 * iqr
upper_whisker = q3 + 1.5 * iqr

# Ausreißer
outliers = data[(data < lower_whisker) | (data > upper_whisker)]

# Boxplot
plt.figure(figsize=(8,5))
plt.boxplot(data, vert=True, patch_artist=True)
plt.title("Boxplot der Kosinus-Ähnlichkeiten")
plt.ylabel("Kosinus-Ähnlichkeit")
plt.show()

print("Q1 (25%):", q1)
print("Median (50%):", median)
print("Q3 (75%):", q3)
print("IQR:", iqr)
print("Ausreißer:", outliers.values)
```
